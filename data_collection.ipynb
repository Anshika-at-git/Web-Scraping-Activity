```python
import requests
```

```python
URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
if res.status_code == 200:
    print(res)

with open("Scraped_data/data2.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
# Beautiful Soup
from bs4 import BeautifulSoup
```

```python
with open("Scraped_data/data2.html") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h1")
```

```python
soup.find_all("h3")
```

```python
# If we need only the content
all_h3 = soup.find_all("h3")

for h3 in all_h3:
    print(h3.get_text(strip = True))
```

```python
# Storing the h3 data
for h3 in all_h3:
    name = h3.get_text(strip = True)
    # print(h3.find_parent("div")) # For all countries parent
    populatiom = h3.find_next("div").select("span.country-population")[0].get_text(strip = True)
```

```python
# Finding the parent of h3 -> Last country
print(h3.find_parent("div"))

# Finding particular attribute of parent
print(h3.find_parent("div")["class"])
```

```python
for h3 in all_h3:
    population = (h3.find_next("div").select_one("span.country-population").get_text(strip = True))
```

```python
# Storing the data

all_countries = []
for h3 in all_h3:
    name = h3.get_text(strip = True)
    populatiom = h3.find_next("div").select("span.country-population")[0].get_text(strip = True)

    all_countries.append([name, population])
```

```python
all_countries
```

```python
import pandas as pd

df = pd.DataFrame(all_countries, columns = ["Name", "Population"])
```

```python
df.to_csv("cleaned_data.csv", index =False)
```
